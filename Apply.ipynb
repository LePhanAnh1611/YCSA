```python
import joblib

model = joblib.load("sentiment.mdl")
vectorizer = joblib.load("vectorizer.pkl")
label_encoder = joblib.load("label_encoder.pkl")

test_comment = input("Gõ bình luận cần phát hiện cảm xúc: ")

transform_comment = vectorizer.transform([test_comment])

pred = model.predict(transform_comment)
sentiment = label_encoder.inverse_transform(pred)

print("Cảm xúc của bình luận là:", sentiment[0])
```
